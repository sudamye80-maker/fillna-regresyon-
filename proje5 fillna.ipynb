```python
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso



df = sns.load_dataset('titanic')



Kullanilacak_sutunlar = ['survived', 'pclass', 'age', 'sibsp', 'parch', 'sex']
df = df[Kullanilacak_sutunlar]


df['sex'] = df['sex'].map({'male': 0, 'female': 1})


df['age'] = df['age'].fillna(df['age'].median())



X = df[['pclass', 'age', 'sibsp', 'parch', 'sex']]
y = df['survived']



X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)



lr = LinearRegression()
lr.fit(X_train, y_train)



ridge = Ridge(alpha=1.0)
ridge.fit(X_train, y_train)



lasso = Lasso(alpha=0.01) 
lasso.fit(X_train, y_train)



print("--- Model R² Başarı Skorları ---")
print(f"Lineer Regresyon Test Skoru : {lr.score(X_test, y_test):.4f}")
print(f"Ridge Regresyon Test Skoru  : {ridge.score(X_test, y_test):.4f}")
print(f"Lasso Regresyon Test Skoru  : {lasso.score(X_test, y_test):.4f}")
print("-" * 32)



katsayilar = pd.DataFrame({
    'Özellik': X.columns,
    'Lineer Reg.': lr.coef_,
    'Ridge (alpha=1.0)': ridge.coef_,
    'Lasso (alpha=0.01)': lasso.coef_
})


print("\n--- Modellerin Katsayı Karşılaştırması ---")
print(katsayilar.to_string(index=False))
```
